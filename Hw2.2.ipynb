```python
python_data = data[data['Мова програмування'] == "Python"]
python_data.head().style.hide(axis="index")
```

```python
print(f"Новий розмір: {python_data.shape[0]} x {python_data.shape[1]}")
```

```python
group_python = python_data.groupby('Посада')[python_data.columns].apply(lambda x: x).iloc[:, 2:]
group_python.head().style.hide(level=1, axis="index")
```

```python
# Не зовсім зрозуміло, що потрібно зробити в цьому завданні. Я зрозумів його якось так.
agg_python = python_data.groupby('Посада')["Зарплата на місяць"].agg(["min", "max"])
agg_python
```

```python
# В цьму завданні також не зрозуміло, що треба зробити. Куди додавати стовпчик "avg"?
# Наскільки я розумію до таблички, котру ми отримали останньою...
def fill_avg_salary(arr: pd.Series):
  return int(arr.mean())

new_data = agg_python
new_data['avg'] = new_data.apply(fill_avg_salary, axis=1)
new_data
```

**Візуалізація**

```python
visual_data = pd.DataFrame({
    "salary": python_data['Зарплата на місяць'],
    "city": python_data['Місто'],
    "company_type": python_data['Тип компанії']
})
plt.figure(figsize=(11, 5))
plt.bar(visual_data['city'], visual_data["salary"], color='y')
plt.title("Середня заробiтня плата по дислокації, серед пайтон-розробників", fontsize = 16)
plt.ylabel("Заробітна плата", fontsize=12)
plt.xlabel("Місто", fontsize=12)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(11, 5))
plt.plot(new_data.index, new_data['avg'], marker='D', linestyle=":")
plt.title("Середня заробiтня плата по посадам, серед пайтон-розробників", color="green", fontsize = 16)
plt.xlabel("Посада", fontsize=12, color="green")
plt.ylabel("Заробітна плата", fontsize=12, color="green")
plt.show()
```

```python
plt.figure(figsize=(11, 5))
plt.bar(visual_data['company_type'], visual_data['salary'], color="green")
plt.title("Заробітна плата в різних типах компаній, серед пайтон-розробників", color="blue", fontsize = 16)
plt.xlabel("Тип компанії", fontsize=12, color="blue")
plt.ylabel("Заробітна плата", fontsize=12, color="blue")
plt.show()
```
